```python
import os

# Force MuJoCo to use the headless EGL rendering driver
os.environ["MUJOCO_GL"] = "egl"

import mujoco
import mediapy as media

# 1. Clear, self-contained XML string (Bypasses external files completely)
xml_scene = """
<mujoco>
  <option gravity="0 0 -9.81"/>
  <worldbody>
    <light pos="0 0 3"/>
    <geom name="floor" type="plane" size="5 5 0.1" rgba="0.5 0.5 0.5 1"/>

    <body name="link1" pos="0 0 1.5">
      <joint name="joint1" type="hinge" axis="0 1 0"/>
      <geom type="capsule" size="0.05" fromto="0 0 0 0 0 -0.4" rgba="1 0 0 1"/>
      <body name="link2" pos="0 0 -0.4">
        <joint name="joint2" type="hinge" axis="0 1 0"/>
        <geom type="capsule" size="0.04" fromto="0 0 0 0 0 -0.4" rgba="0 1 0 1"/>
      </body>
    </body>
  </worldbody>
</mujoco>
"""

# 2. Compile model directly from string
model = mujoco.MjModel.from_xml_string(xml_scene)
data = mujoco.MjData(model)

# 3. Instantiate the 3D offscreen renderer
renderer = mujoco.Renderer(model)

duration = 3.0    # Simulate for 3 seconds
framerate = 60    # 60 frames per second
frames = []

# 4. Run the high-speed computation and capture 3D steps
mujoco.mj_resetData(model, data)
while data.time < duration:
    mujoco.mj_step(model, data)

    # Capture frames at target rendering intervals
    if len(frames) < data.time * framerate:
        renderer.update_scene(data)
        pixels = renderer.render()
        frames.append(pixels)

# 5. Output playable inline movie block
print("Success! Rendering visual player below:")
media.show_video(frames, fps=framerate)
```

```python
import os
# 1. Secure our headless rendering context first
os.environ["MUJOCO_GL"] = "egl"

# 2. Download the official DeepMind model repository directly into your Colab workspace
if not os.path.exists("mujoco_menagerie"):
    print("Cloning the official MuJoCo Menagerie model repository...")
    !git clone --depth 1 https://github.com/google-deepmind/mujoco_menagerie.git
else:
    print("Model repository already downloaded.")

import mujoco
import mediapy as media
import numpy as np

# 3. Load the local file path from the downloaded folder
model_path = "mujoco_menagerie/universal_robots_ur5e/ur5e.xml"
print(f"Compiling the UR5e Robotic Arm model from: {model_path}")

model = mujoco.MjModel.from_xml_path(model_path)
data = mujoco.MjData(model)

# 4. Setup the 3D offscreen renderer
renderer = mujoco.Renderer(model, height=480, width=640)

duration = 4.0    # Run simulation for 4 seconds
framerate = 60    # Video capture rate
frames = []

# 5. Reset simulation states
mujoco.mj_resetData(model, data)

print("Running simulation physics loop...")
while data.time < duration:
    # Send a smooth sine/cosine movement command to the robot's main joints
    data.ctrl[1] = np.sin(data.time * 3.0) * 0.8
    data.ctrl[2] = np.cos(data.time * 2.0) * 0.5

    # Step the physics engine forward
    mujoco.mj_step(model, data)

    # Capture the frame matching our target video framerate
    if len(frames) < data.time * framerate:
        renderer.update_scene(data)
        pixels = renderer.render()
        frames.append(pixels)

print("\nSimulation complete! Compiling video player...")
media.show_video(frames, fps=framerate)
```
